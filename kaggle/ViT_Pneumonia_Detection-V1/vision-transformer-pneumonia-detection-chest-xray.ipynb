## Öz (Abstract)

Vision Transformer (ViT) mimarileri, göğüs radyografisinde (CXR) pnömoni sınıflandırmasında insan-üstü test metrikleri üretebilmektedir; ancak bu metrikler, modelin kararını gerçek patolojik bulgulara mı yoksa veri kümesindeki yapısal yanlılıklara mı dayandırdığı sorusunu yanıtsız bırakır. Bu çalışma, dengeli bir CXR veri kümesi üzerinde ince ayarlanan bir **ViT-B/16** modelinin (Test Doğruluğu **%97.2**, F1 **0.9724**) açıklanabilirlik açısından sistematik bir denetimini sunar ve **kısayol öğrenmesi (shortcut learning / Clever Hans etkisi)** olgusunu üç ardışık fazda teşhis edip hafifletmeyi amaçlar.

İlk fazda, model dikkatinin akciğer parankimi yerine köşe yerleşimli **L/R işaretleyicilerine ve PA/AP çekim artefaktlarına** yöneldiği gösterilir; bu yüzeysel kısayol, merkezi kırpma ve ViT yama-token'ları üzerinde tanımlanan **Right-for-the-Right-Reasons (RRR)** bastırımı ile büyük ölçüde elenir. İkinci fazda, klasik bilgisayarlı görü (Otsu eşikleme) ile çıkarılan akciğer maskelerinin radyoopak pnömoni infiltrasyonlarını "akciğer-dışı" sayarak değerlendirme metriğini geçersiz kıldığı saptanır ve segmentasyon, **U-Net tabanlı derin öğrenme** modeline taşınır. Üçüncü fazda, RRR cezası ile Grad-CAM'in birlikte kullanımının bir **"Silgi Etkisi" (Eraser Effect)** ürettiği gösterilir ve bilgi akışını sadık biçimde izleyen **Attention Rollout**'a geçilir.

Nihai analiz, çalışmanın temel katkısını ortaya koyar: model kenar kısayolundan vazgeçmiş olsa da (anatomik odak oranı **LFR ≈ 0.16** düzeyinde düşük kalmaktadır), dikkati ısrarla **omurga ve mediasten** bölgesinde yoğunlaşmaktadır. Bu örüntünün, ayakta (PA) çekilen sağlıklı görüntüler ile yatağa bağımlı (AP) çekilen ağır hasta görüntüleri arasındaki **postür yanlılığını** sömüren, performansın ardına gizlenmiş bir klinik kısayol olduğu savunulur. Sonuç olarak, geliştirilen XAI çerçevesi, derin öğrenme modellerinin tıbbi teşhiste nasıl sinsi kısayollar üretebileceğini niceliksel olarak kanıtlayan bir denetim aracı olarak konumlandırılır.

## 1. Giriş: Klinik Motivasyon ve Kısayol Öğrenmesi <a id='1'></a>

### 1.1. Problemin Tanımı

Pnömoni, göğüs radyografisinde akciğer parankimindeki **konsolidasyon ve infiltrasyon** alanları olarak görünen, dünya genelinde yüksek mortaliteye sahip bir alt solunum yolu enfeksiyonudur. Derin öğrenme tabanlı bilgisayar destekli teşhis (CAD) sistemleri, bu bulguların otomatik tespitinde dikkat çekici test başarımları rapor etmektedir. Ne var ki, **ayırt edici başarım (discriminative performance) ile klinik güvenilirlik (clinical trustworthiness) aynı şey değildir.** Bir model, doğru sınıf etiketini, görüntüdeki patolojiyle nedensel bir ilişkisi olmayan, yalnızca veri kümesinin toplanma biçiminden kaynaklanan istatistiksel ipuçlarına dayanarak da tahmin edebilir.

### 1.2. Kısayol Öğrenmesi (Shortcut Learning / Clever Hans Etkisi)

Geirhos ve ark. (2020) tarafından biçimsel olarak çerçevelenen **kısayol öğrenmesi**, derin ağların, eğitim dağılımında etiketle güçlü korelasyon gösteren ancak amaçlanan görevi temsil etmeyen "kestirme" karar kurallarını tercih etme eğilimidir. Tıbbi görüntülemede bu olgunun en çarpıcı örneklerinden biri DeGrave ve ark. (2021) çalışmasıdır: COVID-19 tespiti yaptığı düşünülen modellerin, aslında görüntü kaynağına (hastane, çekim cihazı) ve metinsel/pozisyonel artefaktlara baktığı gösterilmiştir. Bu çalışmanın temel hipotezi de buna paraleldir: **CXR pnömoni sınıflandırıcısı, yüksek F1 skorunu büyük olasılıkla en az kısmen kısayollara borçludur ve bu kısayollar dağıtım kayması (distribution shift) altında çökecektir.**

### 1.3. Çalışmanın Evrimi: Katmanlı Bir Kısayol Soğanını Soymak

Bu proje, statik bir mimari tasarımdan ziyade, her aşamada keşfedilen bir Clever Hans yanılgısının teşhis edilip hafifletildiği **iteratif bir teşhis süreci** olarak gelişmiştir. Bu evrim üç faza ayrılır:

> **Faz 1 — Kenar Kısayolları ve Statik RRR.** İlk modeller, dikkati akciğer dokusu yerine köşelerdeki **L/R (Sağ/Sol) işaretleyicilerine** ve **PA/AP çekim geometrisinin** yarattığı kenar dokusu farklarına yöneltmiştir. Çözüm olarak merkezi kırpma (`CenterCrop`) ve köşe yama-token'larını cezalandıran statik bir RRR bastırımı uygulanmış, kenar kısayolu oranı yaklaşık **%100'den %1.7'ye** indirilmiştir. Ancak akciğer-odak oranı (LFR) klinik olarak kabul edilemez biçimde düşük (~%10) kalmıştır.
>
> **Faz 2 — LFR Yanılgısı ve Anatomik Maskeleme.** Düşük LFR'nin kaynağı incelendiğinde, sorunun modelde değil **değerlendirme metodolojisinde** olduğu görülmüştür: klasik Otsu eşikleme + morfoloji, radyoopak (parlak) pnömoni infiltrasyonlarını "akciğer-dışı yoğun doku" sayıp maske dışına itmekteydi. Bu, model doğru yere baksa bile LFR'yi yapay olarak sıfıra çekiyordu. Çözüm: klasik segmentasyonun terk edilip **U-Net tabanlı derin öğrenme segmentasyonuna** geçilmesi.
>
> **Faz 3 — Anatomik RRR ve "Silgi Etkisi".** Eğitim, statik köşelerden dinamik anatomik sınırlara taşınmıştır (`AnatomySuppressionHook`). Ancak Grad-CAM ısı haritaları odağı omurga/kalp üzerinde göstermiştir. Mimari analiz, ViT'te RRR cezası ile Grad-CAM'in birlikte kullanımının bir **"Silgi Etkisi" (Eraser Effect)** ürettiğini ortaya koymuştur (bkz. §6). Bu yanılsamayı aşmak için Grad-CAM analizden çıkarılıp **Attention Rollout**'a geçilmiştir.

## 1.1. Anatomik Önsel: Derin Öğrenme Tabanlı Akciğer Segmentasyonu <a id='1b'></a>

Anatomik olarak yönlendirilmiş RRR'nin temel taşı, modele "nereye **bakmaması** gerektiğini" söyleyen güvenilir bir **akciğer maskesidir**. Bu maskenin kalitesi, hem eğitim sinyalini (akciğer-dışı bastırım) hem de değerlendirme metriğini (LFR) doğrudan belirler.

### Klasik Segmentasyonun Başarısızlığı

İlk yaklaşımda akciğer alanı, Otsu eşikleme ve morfolojik işlemlerle çıkarılmıştır. Bu yöntem temelde bir **yoğunluk (intensity) varsayımına** dayanır: akciğerler havayla dolu olduğundan radyolüsen (koyu) görünür. Ancak pnömoninin radyolojik imzası tam da bu varsayımı çürütür — konsolidasyon ve plevral efüzyon **radyoopak (parlak)** alanlardır. Sonuç olarak Otsu, hastalıklı akciğer bölgelerini "yoğun doku" sayıp maske dışına iter. Bu, paradoksal bir değerlendirme yanılgısı doğurur:

$$\text{Model doğru patolojiye baksa bile} \;\Rightarrow\; L(u_{\text{patoloji}}) = 0 \;\Rightarrow\; \text{LFR} \to 0.$$

Yani metrik, modeli haklı olduğu durumda cezalandırmaktaydı.

### Derin Öğrenme Çözümü

Bu nedenle segmentasyon, yoğunluk önseline değil, geniş ölçekli anotasyonlu CXR verisiyle eğitilmiş **U-Net tabanlı derin öğrenme modellerine** taşınmıştır. Bu modeller akciğer sınırlarını yoğunluktan bağımsız, anatomik bağlamla öğrendiğinden, radyoopak infiltrasyonlar akciğer içinde doğru biçimde kalır. Çalışmada iki ayrı segmentasyon ağı kullanılır ve bu **kasıtlı bir metodolojik tercihtir**:

- **Eğitim zamanı maskeleme** (bu hücrede yüklenen `lungmask` R231 modeli) — RRR cezasını dinamik olarak yönlendirmek için kullanılır.
- **Değerlendirme zamanı doğrulama** (§9'da yüklenen `ianpan/chest-x-ray-basic` modeli) — LFR'yi *bağımsız* bir segmenter üzerinden hesaplayarak döngüsel kanıt (circularity) riskini ortadan kaldırır.

Maske, yalnızca **sağ ve sol akciğer alanlarını** içerecek biçimde optimize edilmiş; kalp, mediasten ve diyafram dışlanmıştır. Bu ayrım, §10'da tartışılan omurga/mediasten kısayolunun tespitinde belirleyici olacaktır.

```python
# %% [markdown]
# # Task 1: Deep Learning Lung Segmentation (Susturulmuş Versiyon)
# !pip install lungmask  # Uncomment to install if not present

import torch
import numpy as np
from lungmask import LMInferer
import warnings
import logging
import os
import contextlib

warnings.filterwarnings("ignore", category=UserWarning, module="lungmask")

# Lungmask'in kendi loglarını sustur (Sadece kritik hataları gösterir)
logging.getLogger("lungmask").setLevel(logging.CRITICAL)

# --- Global Cache for Model ---
_LUNG_SEG_INFERER = None

def get_lung_segmenter():
    """Loads and returns the lung segmentation inferer using the modern API."""
    global _LUNG_SEG_INFERER
    if _LUNG_SEG_INFERER is None:
        print("  [LungSeg] Initializing LMInferer (R231 model)...")
        # İlk yüklemedeki indirme çubuklarını da gizlemek istersen aşağıyı kullanabilirsin.
        # Biz sadece modeli oluşturuyoruz.
        _LUNG_SEG_INFERER = LMInferer()
        print("  [LungSeg] Model ready.")
    return _LUNG_SEG_INFERER

def extract_lung_mask_dl(img_tensor, target_size=224):
    """
    Extracts an accurate lung mask using lungmask's LMInferer.
    Completely silenced to prevent console spam.
    """
    inferer = get_lung_segmenter()
    is_batched = (img_tensor.dim() == 4)
    
    img_cpu = img_tensor.detach().cpu()
    
    mean = torch.tensor(CONFIG["mean"]).view(-1, 1, 1)
    std = torch.tensor(CONFIG["std"]).view(-1, 1, 1)
    
    if is_batched:
        img_denorm = (img_cpu * std + mean).clamp(0, 1)
    else:
        img_denorm = (img_cpu * std + mean).clamp(0, 1).unsqueeze(0)

    B = img_denorm.shape[0]
    masks = []
    
    for i in range(B):
        img_gray = img_denorm[i].mean(dim=0).numpy() * 255
        vol_input = np.expand_dims(img_gray.astype(np.float32), axis=0) 
        
        # SİHİRLİ DOKUNUŞ: Bu blok içindeki tüm console/print (stdout ve stderr) çıktılarını yut!
        with open(os.devnull, 'w') as f, contextlib.redirect_stdout(f), contextlib.redirect_stderr(f):
            segmentation = inferer.apply(vol_input) 
        
        mask_np = (segmentation[0] > 0).astype(np.float32)
        masks.append(mask_np)

    mask_batch_np = np.stack(masks, axis=0) 
    mask_tensor = torch.from_numpy(mask_batch_np).float().unsqueeze(1) 

    if mask_tensor.shape[-1] != target_size:
        mask_tensor = torch.nn.functional.interpolate(
            mask_tensor, size=(target_size, target_size), mode='nearest'
        )
    
    if not is_batched:
        mask_tensor = mask_tensor.squeeze(0) 
        
    return mask_tensor
```

```python
# ── v2 Hata Teşhisi: Skor milyonlarca çıkmasının simülasyonu ──────────
print("v2 Hata Teşhisi: Rollout matris değerleri neden büyüyor?")
print("="*60)

# 197×197 matris (1 CLS + 196 patch) üzerinde örnek
N = 197
# Rastgele attention matrisi (softmax sonrası, gerçekçi)
torch.manual_seed(0)
attn = torch.softmax(torch.randn(N, N), dim=-1)

# v2 hatası: normalize ETMEDEN residual çarpımı yapıldı
result_wrong = torch.eye(N)
for _ in range(12):  # 12 katman
    a = attn + torch.eye(N)
    a = a / a.sum(-1, keepdim=True)
    result_wrong = torch.matmul(a, result_wrong)

cls_wrong = result_wrong[0, 1:]  # CLS→patch
print(f"  v2 (normalize edilmemiş): min={cls_wrong.min():.2e}, max={cls_wrong.max():.2e}")
print(f"  → Değerler çok küçük/büyük, oran hesabı anlamsız hale geliyor!")

# v3 düzeltmesi: her katman sonrası değerleri [0,1]'e normalize et
result_correct = torch.eye(N)
for _ in range(12):
    a = attn + torch.eye(N)
    a = a / a.sum(-1, keepdim=True)
    result_correct = torch.matmul(a, result_correct)
    # ← v3: Her adımda normalize et
    r = result_correct
    r = (r - r.min()) / (r.max() - r.min() + 1e-8)
    result_correct = r

cls_correct = result_correct[0, 1:]
print(f"\n  v3 (adım-adım normalize): min={cls_correct.min():.4f}, max={cls_correct.max():.4f}")
print(f"  → Değerler [0,1] aralığında, oran hesabı anlamlı!")
print()
print("  Sonuç: border/center oranı artık 0-5 arasında çıkacak (milyon DEĞİL)")
```

## 2. Veri Kümesi ve Ön İşleme: Kenar Kısayollarının Eliminasyonu <a id='2'></a>

### 2.1. Yapısal Yanlılığın Kaynağı

CXR veri kümelerinde sınıflar arası dağılım, görüntü içeriğinin ötesinde **çekim protokolüyle** korelasyon gösterir. Bu çalışmada histogram analiziyle (bkz. §2.1) istatistiksel olarak anlamlı ($p < 10^{-4}$) bir örüntü doğrulanmıştır:

- **NORMAL** sınıfı ağırlıkla yetişkin, ayakta **PA (Posterior–Anterior)** çekimlerinden oluşur; köşeler boş (siyah) arka plandır.
- **PNEUMONIA** sınıfı ağırlıkla pediatrik veya ağır hasta **AP (Anterior–Posterior)** çekimlerinden oluşur; köşelerde toraks dokusu taşar.

Buna ek olarak, radyograflara yerleştirilen **L/R (Sağ/Sol) yön işaretleyicileri** köşelerde konumlanır. Bu iki etken birleşince, model için akciğeri hiç incelemeden sınıfı tahmin etmeye yeten bir **çift kenar kısayolu** oluşur.

### 2.2. İki Katmanlı Ön İşleme Hattı

Bu yapısal yanlılığı kaynağında ortadan kaldırmak için aşağıdaki ardışık dönüşüm uygulanır:

$$\text{Resize}(256) \;\rightarrow\; \text{CenterCrop}(c) \;\rightarrow\; \text{Resize}(224) \;\rightarrow\; \text{CornerLetterMask}(r)$$

1. **Merkezi kırpma**, çevresel dokuyu (PA/AP geometri farkını) atarak görüntünün merkez bölgesini korur ve böylece köşe temelli ayrım sinyalini fiziksel olarak yok eder.
2. **`CornerLetterMask`**, kalan kenar şeritlerini düşük varyanslı Gauss gürültüsüyle ($\sigma = 0.03$) doldurur. Sabit bir değerle doldurmak yerine **yumuşak (soft) maskeleme** tercih edilir; bu, ani yoğunluk süreksizliklerinin kendisinin bir kısayol ipucuna dönüşmesini engeller. Maske, normalizasyondan **önce** uygulanır ki gürültü istatistikleri görüntü dağılımıyla tutarlı kalsın.

### 2.3. Dengeli Bölümleme ve Veri Sızıntısı Denetimi

Orijinal doğrulama kümesinin aşırı küçük olması (~30 örnek) F1 tahminini güvenilmez kılmaktadır. Bu nedenle doğrulama ve test havuzları birleştirilip her sınıftan eşit sayıda örnek içeren **tamamen dengeli** alt kümelere ($n_{\text{NORMAL}} = n_{\text{PNEUMONIA}}$) bölünür. Ayrıca dosya imzası (ad + boyut) tabanlı bir kontrol ile **train ↔ val/test** ve **val ↔ test** çakışmaları denetlenerek **veri sızıntısı (data leakage)** sistematik olarak engellenir — bu, raporlanan metriklerin geçerliliği için zorunlu bir adımdır.

```python
# ── CenterCrop + CornerLetterMask Kombinasyonu ───────────────────────
#
# SORUN (histogram analizi ile kanıtlandı):
#   NORMAL    → Yetişkin PA çekim → köşeler SIYAH (boş arka plan)
#   PNEUMONIA → Çocuk AP çekim   → köşelerde DOKU var (toraks taşıyor)
#   p=0.0000 → istatistiksel anlamlı fark → model bu farki kisayol olarak öğreniyor
#   R/L harfleri de köşede → çift kısayol riski
#
# ÇÖZÜM — 2 katmanlı pipeline:
#   1. Resize(256) → CenterCrop(180) → Resize(224)
#      → Köşe dokusunu (PA/AP farkı) tamamen atar
#      → 180/256 = %70 merkez alan korunur
#   2. CornerLetterMask(ratio=0.10)
#      → Kalan kenar artefaktlarını (R/L harfleri) gürültüyle kapatır
#
class CornerLetterMask:
    """
    Kenar şeritlerini düşük Gaussian gürültüyle doldurur.
    CenterCrop(180) sonrası kalan R/L harflerini ve kenar artefaktlarını kör eder.
    ratio=0.10 → ~22px her kenar (224*0.10)
    """
    def __init__(self, ratio=0.10, noise_std=0.03):
        self.ratio     = ratio
        self.noise_std = noise_std

    def __call__(self, tensor):
        C, H, W = tensor.shape
        b = int(H * self.ratio)
        t = tensor.clone()
        noise = torch.randn(C, H, W) * self.noise_std
        
        # DÜZELTME: Köşelerin iki kere ezilmesini engellemek için dilimleme düzeltildi
        t[:, :b, :] = noise[:, :b, :]           # üst şerit (tamamı)
        t[:, -b:, :] = noise[:, -b:, :]         # alt şerit (tamamı)
        t[:, b:-b, :b] = noise[:, b:-b, :b]     # sol şerit (üst ve alt köşeler hariç)
        t[:, b:-b, -b:] = noise[:, b:-b, -b:]   # sağ şerit (üst ve alt köşeler hariç)
        return t

corner_letter_mask = CornerLetterMask(
    ratio     = CONFIG["corner_mask_ratio"],  # 0.10
    noise_std = 0.03
)

# ── Transformasyonlar ─────────────────────────────────────────────────
_crop = CONFIG["center_crop_size"]   # 180
_size = CONFIG["img_size"]           # 224

data_transforms = {
    'train': transforms.Compose([
        transforms.Resize(256),
        transforms.CenterCrop(_crop),          # köşe doku farkını at
        transforms.Resize((_size, _size)),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(degrees=10),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.RandomAffine(degrees=0, translate=(0.05, 0.05)),
        transforms.ToTensor(),
        corner_letter_mask,                    # DÜZELTME: Normalizasyondan ÖNCE eklendi
        transforms.Normalize(CONFIG["mean"], CONFIG["std"]),
    ]),
    'val_test': transforms.Compose([
        transforms.Resize(256),
        transforms.CenterCrop(_crop),
        transforms.Resize((_size, _size)),
        transforms.ToTensor(),
        corner_letter_mask,                    # DÜZELTME: Normalizasyondan ÖNCE eklendi
        transforms.Normalize(CONFIG["mean"], CONFIG["std"]),
    ]),
    'raw': transforms.Compose([   # XAI için — maskelemesiz ama crop var
        transforms.Resize(256),
        transforms.CenterCrop(_crop),
        transforms.Resize((_size, _size)),
        transforms.ToTensor(),
        transforms.Normalize(CONFIG["mean"], CONFIG["std"]),
    ]),
}

# ══════════════════════════════════════════════════════════════════════
# v4 DEĞİŞİKLİK: Dengeli Val/Test Bölümlemesi + Veri Sızıntısı Önleme
# ══════════════════════════════════════════════════════════════════════
#
# SORUN:  Orijinal datasette val çok küçük (~30), test büyük (~1700).
#         Bu dengesizlik val F1'i güvenilmez yapıyor.
#
# ÇÖZÜM:  Val + Test havuzu birleştirilir → eşit ve dengeli bölünür.
#         Her iki sette Normal / Pneumonia sayısı eşit.
#         Dosya yolu kontrolü ile train↔val/test sızıntısı engellenir.
#
from torch.utils.data import Subset
import hashlib

# 1) Train dataseti — dokunma
train_ds = datasets.ImageFolder(TRAIN_PATH, data_transforms['train'])

# 2) Val + Test havuzunu birleştir (aynı transform ile)
val_full   = datasets.ImageFolder(VAL_PATH,  data_transforms['val_test'])
test_full  = datasets.ImageFolder(TEST_PATH, data_transforms['val_test'])

# Tüm örnekleri (dosya_yolu, etiket) olarak topla
pool_samples = val_full.samples + test_full.samples  # list of (path, label)
pool_total   = len(pool_samples)

print(f"{'='*62}")
print(f"  Havuz: Val({len(val_full.samples)}) + Test({len(test_full.samples)}) = {pool_total}")

# ── 3) Veri Sızıntısı (Data Leakage) Kontrolü ────────────────────────
# Dosya adı + boyut hash'i ile train vs havuz çakışması kontrol et
def file_signature(path):
    """Dosya adı + boyut birleşimi → hızlı benzersizlik kontrolü."""
    fname = os.path.basename(path)
    fsize = os.path.getsize(path) if os.path.exists(path) else 0
    return f"{fname}_{fsize}"

train_sigs = set(file_signature(p) for p, _ in train_ds.samples)
pool_sigs  = [file_signature(p) for p, _ in pool_samples]

leak_count = sum(1 for s in pool_sigs if s in train_sigs)
if leak_count > 0:
    print(f"  ⚠️  UYARI: {leak_count} dosya train seti ile çakışıyor!")
    print(f"       Bu dosyalar val/test havuzundan çıkarılıyor...")
    # Çakışan dosyaları havuzdan çıkar
    pool_samples = [(p, l) for (p, l), s in zip(pool_samples, pool_sigs)
                    if s not in train_sigs]
    pool_total = len(pool_samples)
    print(f"       Temizlenmiş havuz: {pool_total}")
else:
    print(f"  ✅ Veri sızıntısı yok — train ile çakışan dosya bulunamadı.")

# ── 4) Sınıf bazında ayır ────────────────────────────────────────────
class_to_idx = train_ds.class_to_idx  # {'NORMAL': 0, 'PNEUMONIA': 1}
idx_normal   = class_to_idx['NORMAL']
idx_pneum    = class_to_idx['PNEUMONIA']

normal_pool  = [(p, l) for p, l in pool_samples if l == idx_normal]
pneum_pool   = [(p, l) for p, l in pool_samples if l == idx_pneum]

print(f"  Havuzda NORMAL: {len(normal_pool)}, PNEUMONIA: {len(pneum_pool)}")

# ── 5) Dengeli bölme: her sette eşit Normal + Pneumonia ──────────────
# Her set = havuz/2.  Her sette min(normal, pneum)/2 kadar her sınıftan al.
# Böylece val ve test tam dengeli olur.
np.random.seed(SEED)

# Karıştır
random.shuffle(normal_pool)
random.shuffle(pneum_pool)

# Her sınıftan her sete düşecek miktar
n_per_class_per_set = min(len(normal_pool), len(pneum_pool)) // 2
print(f"  Her sette sınıf başına: {n_per_class_per_set}")
print(f"  Her set toplam: {n_per_class_per_set * 2}")

# Böl
val_samples  = normal_pool[:n_per_class_per_set] + pneum_pool[:n_per_class_per_set]
test_samples = normal_pool[n_per_class_per_set:2*n_per_class_per_set] + \
               pneum_pool[n_per_class_per_set:2*n_per_class_per_set]

# Karıştır (sınıf sırası bozulsun)
random.shuffle(val_samples)
random.shuffle(test_samples)

# ── 6) Custom Dataset oluştur ─────────────────────────────────────────
from torch.utils.data import Dataset

class BalancedSubset(Dataset):
    """Önceden belirlenmiş (path, label) listesinden dataset oluşturur."""
    def __init__(self, samples, transform=None):
        self.samples   = samples
        self.transform = transform
        self.targets   = [l for _, l in samples]
    
    def __len__(self):
        return len(self.samples)
    
    def __getitem__(self, idx):
        path, label = self.samples[idx]
        img = Image.open(path).convert('RGB')
        if self.transform:
            img = self.transform(img)
        return img, label

image_datasets = {
    'train'   : train_ds,
    'val'     : BalancedSubset(val_samples,  data_transforms['val_test']),
    'test'    : BalancedSubset(test_samples, data_transforms['val_test']),
    'test_raw': BalancedSubset(test_samples, data_transforms['raw']),
}

dataloaders = {
    k: DataLoader(ds, batch_size=CONFIG["batch_size"],
                  shuffle=(k=='train'), num_workers=2, pin_memory=True)
    for k, ds in image_datasets.items()
}

CLASS_NAMES  = train_ds.classes
CLASS_TO_IDX = train_ds.class_to_idx
IDX_TO_CLASS = {v:k for k,v in CLASS_TO_IDX.items()}
CLASS_COLORS = {'NORMAL':'#2E8B57','PNEUMONIA':'#DC143C'}

# ── 7) Doğrulama raporu ──────────────────────────────────────────────
print(f"\n{'='*62}")
print(f"  SON DURUM (Dengeli Bölümleme)")
print(f"{'='*62}")
for split in ['train','val','test']:
    ds = image_datasets[split]
    samples = ds.samples if hasattr(ds, 'samples') else [(None, l) for l in ds.targets]
    labels = [l for _, l in samples]
    cnt = Counter(labels); tot = len(labels)
    n_n = cnt.get(CLASS_TO_IDX['NORMAL'], 0)
    n_p = cnt.get(CLASS_TO_IDX['PNEUMONIA'], 0)
    print(f"  {split.upper():<6}: {tot:>5}  | NORMAL:{n_n:>4}({n_n/tot:.0%}) PNOM:{n_p:>4}({n_p/tot:.0%})")
print(f"{'='*62}")

# ── 8) Çapraz sızıntı kontrolü — val↔test ────────────────────────────
val_paths  = set(p for p, _ in val_samples)
test_paths = set(p for p, _ in test_samples)
cross_leak = val_paths & test_paths
if cross_leak:
    print(f"  ⚠️  Val↔Test çakışma: {len(cross_leak)} dosya!")
else:
    print(f"  ✅ Val↔Test çakışma yok — setler tamamen ayrık.")

# Train↔Val/Test kontrolü (dosya yolu bazlı)
train_paths = set(p for p, _ in train_ds.samples)
tv_leak = train_paths & (val_paths | test_paths)
if tv_leak:
    print(f"  ⚠️  Train↔Val/Test çakışma: {len(tv_leak)} dosya!")
else:
    print(f"  ✅ Train↔Val/Test çakışma yok — veri sızıntısı engellendi.")

print(f"\n  Pipeline: Resize(256) → CenterCrop({_crop}) → Resize(224) → CornerLetterMask({CONFIG['corner_mask_ratio']:.2f})")
print(f"  PA/AP köşe farkı: elendi | R/L harfleri: gürültüyle kapatıldı")
```

```python
# ── Maskeleme Önizleme ─────────────────────────────────────────────────
def denormalize(t, mean=CONFIG["mean"], std=CONFIG["std"]):
    x = t.clone()
    for i,(m,s) in enumerate(zip(mean,std)): x[i] = x[i]*s+m
    return x.clamp(0,1)

def collect_samples(loader, n=5):
    samp = {c:[] for c in CLASS_NAMES}
    for imgs,lbls in loader:
        for img,lbl in zip(imgs,lbls):
            c = IDX_TO_CLASS[lbl.item()]
            if len(samp[c])<n: samp[c].append(img)
        if all(len(v)>=n for v in samp.values()): break
    return samp

samples_raw = collect_samples(dataloaders['test_raw'], n=6)
samples     = collect_samples(dataloaders['test'],     n=6)

fig, axes = plt.subplots(2, 6, figsize=(24, 7))
fig.suptitle('CenterCrop+CornerLetterMask: Ham → Maskeli  (her sütun bir örnek)',
             fontsize=12, fontweight='bold')
for row, cls in enumerate(CLASS_NAMES):
    for col in range(3):
        # Ham
        ax = axes[row, col*2]
        ax.imshow(denormalize(samples_raw[cls][col]).permute(1,2,0).numpy(), cmap='gray')
        ax.set_title(f'Ham #{col+1}', fontsize=8, color=CLASS_COLORS[cls])
        ax.axis('off')
        # Maskeli
        ax = axes[row, col*2+1]
        ax.imshow(denormalize(samples[cls][col]).permute(1,2,0).numpy(), cmap='gray')
        ax.set_title(f'Maskeli #{col+1}', fontsize=8, color='navy')
        ax.axis('off')
        b = int(224*CONFIG["corner_mask_ratio"])
        for (r0,c0) in [(0,0),(0,224-b),(224-b,0),(224-b,224-b)]:
            axes[row,col*2+1].add_patch(patches.Rectangle((c0,r0),b,b,
                lw=1.5,edgecolor='lime',facecolor='none',linestyle='--'))
plt.tight_layout()
plt.savefig('fig_01_masking.png',dpi=130,bbox_inches='tight')
plt.show()
```

```python
# ── R/L Harf + PA/AP Kısayol Analizi ─────────────────────────────────
# CenterCrop öncesi/sonrası köşe dağılımı karşılaştır

def analyze_edge_by_class(loader_raw, n=30):
    b = int(224 * 0.15)  # eski oran — karşılaştırma için
    edge_cls  = {c: [] for c in CLASS_NAMES}
    center_cls = {c: [] for c in CLASS_NAMES}
    cnt = 0
    for imgs, lbls in loader_raw:
        for img, lbl in zip(imgs, lbls):
            if cnt >= n: break
            cls    = IDX_TO_CLASS[lbl.item()]
            img_np = denormalize(img).mean(0).numpy()
            edge_pixels = np.concatenate([
                img_np[:b, :].ravel(), img_np[-b:, :].ravel(),
                img_np[:, :b].ravel(), img_np[:, -b:].ravel(),
            ])
            edge_cls[cls].append(edge_pixels.mean())
            center_cls[cls].append(img_np[b:-b, b:-b].mean())
            cnt += 1
        if cnt >= n: break
    return edge_cls, center_cls

print("Kenar bolge analizi (CenterCrop sonrasi)...")
edge_d, center_d = analyze_edge_by_class(dataloaders['test_raw'], n=30)

from scipy import stats as _sp
t_e, p_edge   = _sp.ttest_ind(edge_d['NORMAL'],   edge_d['PNEUMONIA'])
t_c, p_center = _sp.ttest_ind(center_d['NORMAL'], center_d['PNEUMONIA'])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('CenterCrop+CornerLetterMask Sonrasi Kenar Analizi\n'
             'p<0.05 → hala kisayol riski var', fontsize=12, fontweight='bold')

for ax, data, title in [
    (axes[0], edge_d,   f'Kenar Bolge  (p={p_edge:.4f}) {"RISK" if p_edge<0.05 else "TEMIZ"}'),
    (axes[1], center_d, f'Merkez Bolge (p={p_center:.4f})'),
]:
    for cls in CLASS_NAMES:
        ax.hist(data[cls], bins=10, alpha=0.65,
                color=CLASS_COLORS[cls], label=cls, density=True, edgecolor='white')
    ax.set_title(title); ax.legend(); sns.despine(ax=ax)

plt.tight_layout()
plt.savefig('fig_edge_after_crop.png', dpi=120, bbox_inches='tight')
plt.show()
print(f"Kenar p={p_edge:.4f} → {'HALA KISAYOL RISKI' if p_edge<0.05 else 'TEMIZ - kisayol yok'}")
print(f"Merkez p={p_center:.4f} → klinik bilgi")
```

## 3. ViT-B/16 Mimarisi ve Yama-Token Dikkat Akışı <a id='4'></a>

### 3.1. Tokenizasyon

Vision Transformer (Dosovitskiy ve ark., 2021), görüntüyü örtüşmeyen yamalara böler ve her yamayı bir **token** olarak doğrusal gömme uzayına yansıtır. $224 \times 224$ girdi ve $16 \times 16$ yama için:

$$N_p = \left(\frac{224}{16}\right)^2 = 14 \times 14 = 196 \;\text{yama-token} \;+\; 1 \;\text{sınıflandırma (CLS) token} \;=\; 197 \;\text{token}.$$

Her token, $D = 768$ boyutlu bir gömme vektörü $\mathbf{z}_p \in \mathbb{R}^{768}$ ile temsil edilir. Sınıflandırma kararı, çok-katmanlı öz-dikkat (self-attention) yığını boyunca diğer tüm token'lardan bilgi toplayan CLS token üzerinden verilir.

### 3.2. RRR İçin Kritik Sonuç

ViT'in token-temelli yapısı, anatomik düzenlileştirme için **doğal bir mekanizma** sağlar: her token uzamsal olarak tanımlı bir yamaya karşılık geldiğinden, "akciğer-dışı" yamalar doğrudan adreslenebilir. Köşe yamaları (örn. $14 \times 14$ ızgarada indeksler $\{0,13,182,195\}$) tıbbi bilgi taşımaz. Bu görselleştirme, köşe token konumlarını ızgara üzerinde işaretleyerek §4'teki anatomik bastırımın hangi token'lar üzerinde çalıştığını somutlaştırır.

### 3.3. İnce Ayar Stratejisi

Hesaplama verimliliği ve aşırı-öğrenmeyi sınırlamak adına, ImageNet ön-eğitimli ağırlıklar dondurulur; yalnızca **son iki kodlayıcı bloğu**, son katman normalizasyonu (LayerNorm) ve iki sınıflı yeni sınıflandırma başlığı eğitilir. Bu, transfer öğrenmede yaygın bir parametre-verimli ince ayar yaklaşımıdır.

## 4. Anatomik Olarak Yönlendirilmiş RRR ile Model Eğitimi <a id='5'></a>

### 4.1. Right for the Right Reasons (RRR)

Ross ve ark. (2017), bir modelin yalnızca **doğru tahmini** değil, **doğru gerekçeyi** de öğrenmesini sağlamak için açıklamayı (model gradyanını) düzenlileştiren RRR çerçevesini önermiştir. Özgün formülasyonda, alan-uzmanı tarafından "ilgisiz" işaretlenen girdi bölgelerine ($A_{nd} = 1$) ait girdi-gradyanları cezalandırılır:

$$\mathcal{L}_{\text{RRR}}^{\text{Ross}} = \sum_{n=1}^{N} \sum_{d=1}^{D} \left( A_{nd} \, \frac{\partial}{\partial x_{nd}} \sum_{k=1}^{K} \log \hat{y}_{nk} \right)^{2}.$$

### 4.2. Bu Çalışmanın Uyarlaması: Akciğer-Dışı Token Bastırımı

Çalışmada, hesaplama açısından ağır olan girdi-gradyan cezası yerine, ViT'in yapısını sömüren bir **gömme-uzayı (embedding-space) bastırım varyantı** kullanılır. `AnatomySuppressionHook`, kodlayıcı çıkışını yakalar ve akciğer-dışı yamaların gömme normunu doğrudan cezalandırır. Düzenlileştirme sinyali şu adımlarla anatomik maskeden türetilir:

**(i) Yama-ızgarasına indirgeme.** U-Net akciğer maskesi $\mathbf{M} \in \{0,1\}^{224 \times 224}$, $16 \times 16$ ortalama-havuzlama ile yama ızgarasına taşınır:

$$\overline{S}_{p} = \frac{1}{256} \sum_{(a,b) \in \mathcal{P}_p} \mathbf{M}(a,b), \qquad p = 1, \dots, 196.$$

**(ii) Akciğer-dışı maske.** Bir yama, alanının %10'undan azı akciğerse "akciğer-dışı" sayılır:

$$M_{p}^{\text{nl}} = \mathbb{1}\!\left[\, \overline{S}_{p} < \tau_{\text{lung}} \,\right], \qquad \tau_{\text{lung}} = 0.10.$$

**(iii) Anatomik ceza.** Akciğer-dışı yamaların gömme enerjisi, parti ve token üzerinden ortalanarak cezalandırılır ($\mathbf{z}_{i,p} \in \mathbb{R}^{768}$ kodlayıcı çıkışıdır):

$$\mathcal{L}_{\text{anat}} = \frac{1}{B \, N_p} \sum_{i=1}^{B} \sum_{p=1}^{N_p} M_{i,p}^{\text{nl}} \, \bigl\lVert \mathbf{z}_{i,p} \bigr\rVert_2^2 .$$

($M_{i,p}^{\text{nl}} \in \{0,1\}$ olduğundan $\bigl\lVert M\,\mathbf{z}\bigr\rVert^2 = M\,\lVert\mathbf{z}\rVert^2$ özdeşliği geçerlidir; ceza yalnızca akciğer-dışı token'larda etkindir.) Kritik tasarım kararı, kancanın kodlayıcı çıkışını **`detach` etmemesidir**: gradyan akışının korunması, cezanın geriye yayılımda gerçekten ağırlıkları güncellemesi için zorunludur.

### 4.3. Toplam Kayıp Fonksiyonu

Model, çok-görevli bir hedefle eğitilir: doğru sınıfı tahmin etmek (çapraz entropi) **ve** kararını anatomik olarak meşru bölgelere dayandırmak (anatomik ceza):

$$\boxed{\;\mathcal{L}_{\text{total}} \;=\; \underbrace{-\frac{1}{B}\sum_{i=1}^{B}\sum_{c=1}^{K} y_{i,c}\,\log \hat{p}_{i,c}}_{\mathcal{L}_{\text{CE}}\;\text{(sınıflandırma)}} \;+\; \lambda \cdot \underbrace{\frac{1}{B\,N_p}\sum_{i,p} M_{i,p}^{\text{nl}}\,\lVert \mathbf{z}_{i,p}\rVert_2^2}_{\mathcal{L}_{\text{anat}}\;\text{(anatomik düzenlileştirme)}} \;}$$

burada $\lambda = 5.0$ ödünleşim katsayısı, $K = 2$ sınıf sayısı, $B$ parti boyutudur.

### 4.4. Optimizasyon Protokolü

Eğitim, ağırlık çürümeli **AdamW** ($\eta = 10^{-4}$, $\text{wd} = 10^{-2}$) ile, **kosinüs tavlama (cosine annealing)** çizelgesi altında 15 dönem boyunca yürütülür; kararlılık için gradyan normu 1.0'a kırpılır. Her parti için akciğer maskesi U-Net ile **dinamik olarak** hesaplanır, dolayısıyla düzenlileştirme sinyali görüntüye özgüdür.

> **İleriye dönük gözlem (bkz. §10).** Eğitim loglarında anatomik ceza $\mathcal{L}_{\text{anat}}$, başlangıçtaki $\sim 3.60$ değerinden son dönemde $\sim 3 \times 10^{-4}$ düzeyine inmektedir. Modelin cezayı sayısal olarak neredeyse sıfırlamasına rağmen (§9'da görüleceği gibi) hâlâ akciğer-dışı bölgelere bakması, $\lambda = 5.0$ katsayısının güçlü bir kısayolu caydırmaya **yetmediğinin** ilk işaretidir.

```python
# ── Model Kurulumu ────────────────────────────────────────────────────
def build_vit(fine_tune_last_n=2):
    model = models.vit_b_16(weights=models.ViT_B_16_Weights.IMAGENET1K_V1)
    for p in model.parameters(): p.requires_grad = False
    for p in model.encoder.layers[-fine_tune_last_n:].parameters(): p.requires_grad = True
    for p in model.encoder.ln.parameters(): p.requires_grad = True
    in_f = model.heads.head.in_features
    model.heads.head = nn.Sequential(
        nn.Dropout(CONFIG["dropout"]),
        nn.Linear(in_f, CONFIG["num_classes"])
    )
    return model.to(device)

model = build_vit(fine_tune_last_n=2)
total_p = sum(p.numel() for p in model.parameters())
train_p = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"  Toplam: {total_p:,} | Eğitilebilir: {train_p:,} ({train_p/total_p:.1%})")
```

```python
class AnatomySuppressionHook:
    """
    ViT embedding'lerini akciğer dışı bölgelerde (non-lung) cezalandırır.
    (Diferansiyel Versiyon - Gradyan akışına izin verir)
    """
    def __init__(self, model):
        self._enc_output = None
        self._non_lung_mask = None
        self._hook = model.encoder.register_forward_hook(self._capture_output)

    def _capture_output(self, module, input, output):
        # DETACH ETMİYORUZ! Geri yayılım (backward pass) için gradyan akışı şart.
        self._enc_output = output 

    def get_penalty(self):
        """
        Eğitim döngüsünde (train_epoch) belirlenen _non_lung_mask ve
        yakalanan _enc_output kullanılarak ceza skorunu hesaplar.
        """
        if self._enc_output is None or self._non_lung_mask is None:
            # Hata vermemesi için gradyanı olan 0.0 döndür
            dummy_device = self._enc_output.device if self._enc_output is not None else 'cpu'
            return torch.tensor(0.0, device=dummy_device, requires_grad=True)
        
        # _enc_output shape: (B, 197, 768)  (1 CLS + 196 patch)
        # _non_lung_mask shape: (B, 14, 14)
        B, N, D = self._enc_output.shape
        
        # Sadece 196 patch'i al (0. indeksteki CLS token'ı atla)
        patch_embeddings = self._enc_output[:, 1:, :]  # (B, 196, 768)
        
        # Maskeyi (B, 14, 14)'ten (B, 196) boyutuna düzleştir
        non_lung_flat = self._non_lung_mask.view(B, -1)  # (B, 196)
        
        # Maskeyi embedding boyutuyla çarpabilmek için (B, 196, 1) yap
        # Akciğer DIŞI olanlar (1.0) kalır, akciğer İÇİ olanlar (0.0) sıfırlanır
        penalty_norm_sq = (patch_embeddings * non_lung_flat.unsqueeze(-1)).pow(2).sum(dim=-1)  # (B, 196)
        
        # Tüm batch ve patch'ler üzerinden ortalama cezayı al
        penalty = penalty_norm_sq.mean()  
        return penalty

    def clear(self):
        # Her batch bitiminde temizlik
        self._enc_output = None
        self._non_lung_mask = None

    def remove(self):
        self._hook.remove()
```

```python
# ── Eğitim Döngüsü — Corner Suppression ──────────────────────────────
optimizer = optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=CONFIG["lr"], weight_decay=CONFIG["weight_decay"]
)
scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=CONFIG["epochs"], eta_min=1e-6
)

def train_epoch_dynamic(model, loader, anatomy_hook, optimizer, criterion_ce, rrr_lambda, device):
    """
    Training loop with anatomically-guided RRR loss.
    
    For each batch:
        1. Get lung masks using DL model.
        2. Downsample masks to 14x14 patch grid.
        3. Define a patch as "non-lung" if it contains < threshold% lung tissue.
        4. Apply the penalty via the anatomy hook.
        5. Total Loss = CE Loss + rrr_lambda * Anatomy Penalty
    """
    model.train()
    total_loss_sum = 0.0
    ce_loss_sum = 0.0
    anatomy_penalty_sum = 0.0
    all_preds = []
    all_labels = []
    lung_tissue_threshold = 0.10 # A patch is "non-lung" if <10% of its area is lung

    for imgs, labels in loader:
        imgs, labels = imgs.to(device), labels.to(device)
        batch_size = imgs.shape[0]
        
        # --- 1. Get Anatomical Lung Masks using Deep Learning Model ---
        # This returns (B, 1, H, W) mask on CPU. Move to GPU.
        lung_masks = extract_lung_mask_dl(imgs, target_size=CONFIG["img_size"]).to(device) # (B, 1, 224, 224)
        
        # --- 2. Downsample mask to ViT's patch grid (14x14) ---
        # Average pooling over 16x16 patches (224/16 = 14)
        patch_mask = torch.nn.functional.avg_pool2d(lung_masks, kernel_size=16, stride=16) # (B, 1, 14, 14)
        patch_mask = patch_mask.squeeze(1) # (B, 14, 14)
        
        # --- 3. Create non_lung_mask: 1 if patch has less than 10% lung tissue ---
        # If a patch is mostly non-lung, we want to penalize the model for looking at it.
        non_lung_mask = (patch_mask < lung_tissue_threshold).float() # (B, 14, 14)
        
        # --- 4. Forward Pass with RRR Hook ---
        optimizer.zero_grad()
        anatomy_hook.clear()
        
        # Pass the non_lung_mask to the hook BEFORE or AFTER forward? 
        # The hook needs the mask to compute the penalty. We'll set it on the hook
        # but the penalty will be computed after the forward pass.
        # Let's pass the mask to the hook as an attribute.
        anatomy_hook._non_lung_mask = non_lung_mask # A bit hacky, better to store it in the hook class
        outputs = model(imgs) # Forward pass (hook captures encoder output)
        
        # --- 5. Calculate Losses ---
        ce_loss = criterion_ce(outputs, labels)
        # Compute the differentiable penalty using the hook's captured output and mask
        # We need to modify the hook to have a compute_penalty method.
        # Let's update the AnatomySuppressionHook class.
        
        # Assuming the hook has a `get_penalty` method that uses self._enc_output
        # and the stored mask, we call it.
        anatomy_penalty = anatomy_hook.get_penalty() 
        
        total_loss = ce_loss + rrr_lambda * anatomy_penalty
        
        # --- 6. Backward Pass ---
        total_loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        
        # --- 7. Logging ---
        total_loss_sum += total_loss.item() * batch_size
        ce_loss_sum += ce_loss.item() * batch_size
        anatomy_penalty_sum += anatomy_penalty.item() * batch_size
        all_preds.extend(outputs.argmax(dim=1).cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        
        anatomy_hook.clear() # Reset captured output for next batch

    n_samples = len(loader.dataset)
    epoch_f1 = f1_score(all_labels, all_preds, zero_division=0)
    
    return (total_loss_sum / n_samples, 
            ce_loss_sum / n_samples, 
            anatomy_penalty_sum / n_samples, 
            epoch_f1)



def validate(model, loader):
    model.eval()
    loss_s, preds_all, labels_all, probs_all = 0., [], [], []
    with torch.no_grad():
        for imgs, lbls in loader:
            imgs, lbls = imgs.to(device), lbls.to(device)
            out   = model(imgs)
            prob  = torch.softmax(out, 1)
            loss_s += criterion_ce(out, lbls).item() * imgs.size(0)
            preds_all.extend(out.argmax(1).cpu().numpy())
            labels_all.extend(lbls.cpu().numpy())
            probs_all.extend(prob[:,1].cpu().numpy())
    N  = len(loader.dataset)
    return (loss_s/N,
            f1_score(labels_all, preds_all, zero_division=0),
            (np.array(preds_all)==np.array(labels_all)).mean(),
            labels_all, preds_all, probs_all)


# --- Initialize the Hook and Run Training ---
# Replace your old training loop with this
anatomy_hook = AnatomySuppressionHook(model)
criterion_ce = nn.CrossEntropyLoss()

print(f"{'='*72}")
print(f"  ANATOMY-GUIDED RRR TRAINING | lambda={CONFIG['rrr_lambda']} | batch={CONFIG['batch_size']} | {CONFIG['epochs']} Epoch")
print(f"{'='*72}")

# Re-initialize history and best model tracking
history = {k:[] for k in ['tr_loss','ce_loss','anatomy_pen','vl_loss','tr_f1','vl_f1']}
best_f1, best_wt = 0.0, None

for epoch in range(CONFIG["epochs"]):
    tr_loss, ce_loss, anat_pen, tr_f1 = train_epoch_dynamic(
        model, dataloaders['train'], anatomy_hook, optimizer, criterion_ce, 
        CONFIG['rrr_lambda'], device
    )
    
    # Validation (re-use your existing validate function)
    vl_loss, vl_f1, _, _, _, _ = validate(model, dataloaders['val']) # Assuming your validate function returns these
    
    # Store history
    history['tr_loss'].append(tr_loss); history['ce_loss'].append(ce_loss)
    history['anatomy_pen'].append(anat_pen); history['vl_loss'].append(vl_loss)
    history['tr_f1'].append(tr_f1); history['vl_f1'].append(vl_f1)
    
    # Save best model
    star = ' ★' if vl_f1 > best_f1 else ''
    if vl_f1 > best_f1:
        best_f1, best_wt = vl_f1, {k:v.clone() for k,v in model.state_dict().items()}
    
    print(f"Ep{epoch+1:3d} | TR Loss:{tr_loss:.4f} | CE:{ce_loss:.4f} | AnatPen:{anat_pen:.6f} | VL F1:{vl_f1:.4f}{star}")

# Load best model after training
if best_wt: model.load_state_dict(best_wt)
print(f"\n★ Best Val F1: {best_f1:.4f} loaded.")
```

## 6. Açıklanabilirlik: Attention Rollout ve "Silgi Etkisi" <a id='7'></a>

### 6.1. Neden Grad-CAM Değil? — "Silgi Etkisi" (Eraser Effect)

Bu çalışmanın metodolojik açıdan en önemli keşiflerinden biri, **RRR cezası ile Grad-CAM'in birlikte kullanıldığında üretilen bir yorumlama yanılsamasıdır.** Anatomik bastırımla eğitilen modelde Grad-CAM ısı haritaları, beklenenin aksine, odağı akciğer loblarında değil **omurga ve kalp üzerinde** göstermiştir. Mimari analiz, bunun bir sinyal değil bir **artefakt** olduğunu ortaya koymuştur:

> Grad-CAM, MLP bloklarındaki aktivasyon-gradyan çarpımını görselleştirir. Anatomik ceza altında model, akciğer-dışı bölgelerin (omurga/kalp) gömme normlarını **bastırmak (sıfırlamak)** için bu bloklarda yoğun "negatif efor" harcar. Grad-CAM, modelin *patolojiyi nerede aradığını* değil, *cezadan kaçmak için neyi sildiğini* yüksek aktivasyon olarak işaretler. Bu yüzden silinen bölgeler paradoksal biçimde "sıcak" görünür.

Bu optik illüzyon, RRR + Grad-CAM kombinasyonunda yanıltıcı bir akciğer-dışı odak görüntüsü üretir. Çözüm, gradyan-temelli bir vekil yerine, bilginin ağ içindeki **gerçek akış yollarını** izleyen bir yönteme — Attention Rollout — geçmektir.

### 6.2. Attention Rollout

Abnar & Zuidema (2020) tarafından önerilen Attention Rollout, CLS token'a ulaşan bilgiyi, tüm katmanlardaki dikkat matrislerinin çarpımı olarak modelleyerek niceler. $L$ katmanlı bir ViT'te, $l$. katmanın başlar-üstü ortalanmış dikkat matrisi $\tilde{A}^{(l)} = \frac{1}{H}\sum_{h} A_h^{(l)}$ olsun. Artık bağlantıları (residual) hesaba katmak için birim matris eklenir ve satır-normalize edilir:

$$\hat{A}^{(l)} = D^{-1}\bigl(\tilde{A}^{(l)} + I\bigr), \qquad D = \mathrm{diag}\!\Bigl(\bigl(\tilde{A}^{(l)} + I\bigr)\mathbf{1}\Bigr).$$

Tüm katmanlar boyunca bilgi akışı, bu matrislerin ardışık çarpımıdır:

$$R \;=\; \prod_{l=L}^{1} \hat{A}^{(l)} \;=\; \hat{A}^{(L)} \hat{A}^{(L-1)} \cdots \hat{A}^{(1)} .$$

Görselleştirme için CLS satırının yama bileşenleri ($R_{0,\,1:}$) alınır, $14 \times 14$ ızgaraya yeniden şekillendirilir ve [0,1] aralığına ölçeklenir. Bu uygulamada iki sayısal incelik vardır: (i) yalnızca **son çıktı** min–maks normalize edilir (ara adımlarda zemin değerinin yapay şişmesi önlenir), (ii) **agresif gürültü eşikleme** ($\text{discard}$) ile yalnızca yüksek-enerjili dikkat korunur.

### 6.3. Saliency Üretimi

`overlay_heatmap` işlevi, üretilen [0,1] haritasını klinik-okunabilirlik için özel bir renk skalasıyla orijinal görüntü üzerine bindirir. `ViTGradCAMV5` sınıfı, karşılaştırma amacıyla korunur; ancak softmax doygunluğundan kaynaklanan **kaybolan gradyan (vanishing gradient)** sorununu aşmak için logit-üzerinden ve diferansiyel (sınıflar-arası fark) modunda çalışır, gradyan tamamen sönerse Attention Rollout'a geri düşer.

```python
MEDICAL_CMAP = LinearSegmentedColormap.from_list(
    'medical', ['#000033','#0033CC','#00CCFF','#FFFF00','#FF4500'], N=256)

def overlay_heatmap(img_t, heatmap, alpha=0.45):
    img_np  = denormalize(img_t).permute(1,2,0).numpy()
    colored = MEDICAL_CMAP(heatmap)[:,:,:3]
    return np.clip((1-alpha)*img_np + alpha*colored, 0, 1), img_np


class AttentionRolloutV4:
    """
    v4 — İki temel düzeltme:

    1. Adım-adım normalize KALDIRILDI:
       Her adımda normalize etmek → harita zemin değerlerini
       yapay yükseltiyordu → border/center median dengeli çıkıyordu.
       Doğru yol: sadece SON çıkışı normalize et.

    2. discard_ratio 0.9 → 0.95:
       Daha agresif gürültü temizleme → sadece gerçek dikkat kalır.
    """
    def __init__(self, model):
        self.model     = model
        self.attn_maps = []
        self._hooks    = []
        self._register()

    def _register(self):
        for layer in self.model.encoder.layers:
            sa   = layer.self_attention
            orig = sa.forward
            def make_patched(orig_fwd):
                def patched(self_mod, *a, **kw):
                    kw['need_weights'] = True
                    return orig_fwd(*a, **kw)
                return patched
            sa.forward = types.MethodType(make_patched(orig), sa)
            def make_hook(obj):
                def hook(m, inp, out):
                    if out[1] is not None:
                        obj.attn_maps.append(out[1].detach().cpu())
                return hook
            self._hooks.append(sa.register_forward_hook(make_hook(self)))

    def remove(self):
        for h in self._hooks: h.remove()

    def compute(self, img_tensor, head_fusion='mean', discard_ratio=0.0):
        self.attn_maps = []
        self.model.eval()
        with torch.no_grad():
            self.model(img_tensor.unsqueeze(0).to(device))
        if not self.attn_maps:
            raise RuntimeError("Attention yakalanamadi.")

        N = self.attn_maps[0].size(-1)
        result = torch.eye(N)

        for attn in self.attn_maps:
            a = attn.squeeze(0)
            if   head_fusion == 'mean': a = a.mean(0)
            elif head_fusion == 'max' : a = a.max(0)[0]
            else                      : a = a.min(0)[0]

            # Agresif gürültü temizleme (0.95)
            flat = a.view(-1)
            thr  = flat.kthvalue(max(1, int(flat.size(0) * discard_ratio)))[0]
            a[a < thr] = 0.0

            a = a + torch.eye(N)
            a = a / (a.sum(-1, keepdim=True) + 1e-8)
            result = torch.matmul(a, result)
            # ← v4: Adım-adım normalize YOK (zemin şişmesi engellendi)

        # Sadece son adımda normalize
        mask = result[0, 1:].numpy()
        n    = int(round(mask.size ** 0.5))
        mask_14 = mask.reshape(n, n)
        mask_14_norm = (mask_14 - mask_14.min()) / (mask_14.max() - mask_14.min() + 1e-8)

        mask_224 = cv2.resize(mask_14_norm.astype(np.float32), (224, 224))
        mask_224 = (mask_224 - mask_224.min()) / (mask_224.max() - mask_224.min() + 1e-8)

        return mask_224, mask_14_norm   # (224x224, 14x14) — her ikisi [0,1]

try: rollout.remove()
except: pass
rollout = AttentionRolloutV4(model)
print(f"Attention Rollout v4 hazir — Son-adim-normalize, discard=0.95")
print(f"   {len(rollout._hooks)} katman izleniyor")
```

```python
class ViTGradCAMV5:
    """
    v5 — Vanishing gradient sorunu kesin çözümü.
 
    v4 sorunu: softmax üzerinden backward; model çok güvenli olunca
              (prob ~ 0.9999) softmax doyma bölgesinde → gradyan ≈ 0
              → CAM tamamen siyah → LFR yapay olarak 0.
 
    v5 yaklaşımı:
      • Backward DOĞRUDAN logit üzerinden (softmax YOK)
      • differential=True modu: (logit_target - logit_other)
        → sınıflar arası ayırt edici özelliklere bakar
      • CAM enerjisi pratikte sıfırsa 'vanished' bayrağı dönülür
        → çağıran taraf fallback yapsın
    """
    def __init__(self, model, target_layer_idx=-1):
        self.model = model
        self.acts  = None
        self.grads = None
        tgt = model.encoder.layers[target_layer_idx].mlp
        self._h1 = tgt.register_forward_hook(
            lambda m, i, o: setattr(self, 'acts', o.detach()))
        self._h2 = tgt.register_full_backward_hook(
            lambda m, gi, go: setattr(self, 'grads', go[0].detach()))
 
    def remove_hooks(self):
        try:
            self._h1.remove(); self._h2.remove()
        except Exception:
            pass
 
    def compute(self, img_tensor, target_class=None, differential=True):
        self.model.eval()
        # Hook'ların tetiklenmesi için parametreler grad gerektirmeli
        for p in self.model.parameters():
            p.requires_grad_(True)
 
        self.acts = None
        self.grads = None
        img = img_tensor.unsqueeze(0).to(device)
 
        output = self.model(img)                # (1, num_classes)
        if target_class is None:
            target_class = output.argmax(1).item()
 
        # === KRITIK DÜZELTME: LOGIT üzerinden backward (softmax YOK) ===
        if differential and output.shape[1] >= 2:
            other = 1 - target_class                # binary varsayımı
            score = output[0, target_class] - output[0, other]
        else:
            score = output[0, target_class]
 
        self.model.zero_grad()
        score.backward()
 
        if self.grads is None:
            raise RuntimeError("Gradyan yakalanamadi (hook tetiklenmedi).")
 
        grads = self.grads[0, 1:, :]   # (196, 768) — CLS hariç
        acts  = self.acts[0,  1:, :]   # (196, 768)
 
        weights = grads.mean(0)                        # (768,)
        cam     = F.relu((acts * weights).sum(-1))     # (196,)
        cam_np  = cam.detach().cpu().numpy().astype(np.float32)
 
        cam_max = float(cam_np.max())
        n       = int(round(cam_np.size ** 0.5))
        cam_224 = cv2.resize(cam_np.reshape(n, n), (224, 224))
 
        # Sadece anlamlı enerji varsa normalize et
        if cam_max > 1e-6:
            cam_224 = (cam_224 - cam_224.min()) / (cam_224.max() - cam_224.min() + 1e-8)
            vanished = False
        else:
            cam_224 = np.zeros_like(cam_224)
            vanished = True
 
        return cam_224, target_class, {'max': cam_max, 'vanished': vanished}
 
 
# Eski v4 instance varsa temizle, v5 örneğini oluştur
try:
    vit_gradcam.remove_hooks()
except Exception:
    pass
 
vit_gradcam = ViTGradCAMV5(model, target_layer_idx=-1)
print("GradCAM v5 hazir — logit backward + differential mode")
print("   (softmax doyumu / vanishing gradient sorunu cozuldu)")
```

## 7. Kısayol Niceliği: İstatistiksel Skorlama <a id='8'></a>

Dikkat haritalarını **niceliksel ve istatistiksel** bir kısayol kararına dönüştürmek için, alan-ağırlıklı bir köşe–merkez oranı tanımlanır. $14 \times 14$ ızgarada, her köşeden $2 \times 2$ olmak üzere toplam 16 yama "köşe kümesi" ($\mathcal{C}$), kalan 180 yama "merkez kümesi" ($\bar{\mathcal{C}}$) olarak ayrılır. Dikkat haritası toplama bölünerek normalize edilir ($\hat{a}_p = a_p / \sum_q a_q$) ve kısayol skoru, iki bölgenin ortalama normalize dikkatinin oranı olarak hesaplanır:

$$\mathrm{SC} \;=\; \frac{\bar{a}_{\mathcal{C}}}{\bar{a}_{\bar{\mathcal{C}}}}, \qquad \bar{a}_{\mathcal{C}} = \frac{1}{|\mathcal{C}|}\sum_{p \in \mathcal{C}} \hat{a}_p, \quad \bar{a}_{\bar{\mathcal{C}}} = \frac{1}{|\bar{\mathcal{C}}|}\sum_{p \in \bar{\mathcal{C}}} \hat{a}_p .$$

Tekdüze (uniform) dikkat dağılımı için $\mathrm{SC} = 1.0$ olur. Skor, parametrik olmayan, sağlam bir hipotez testiyle desteklenir: köşe ve merkez dikkat değerlerinin dağılımları **tek-taraflı Mann–Whitney U testi** ile karşılaştırılır. Kısayol kararı şu kuralla verilir:

$$\text{kısayol} \;=\; (\mathrm{SC} > 1.55) \;\wedge\; (p_{\text{MWU}} < 0.05),$$

ve risk seviyesi $\mathrm{SC}$ değerine göre **DÜŞÜK** ($\le 1.3$), **ORTA** ($1.3 < \mathrm{SC} \le 1.6$) veya **YÜKSEK** ($> 1.6$) olarak sınıflandırılır. Kalibrasyon eşiği 1.55, yaklaşık %5 yanlış-pozitif oranını hedefleyecek biçimde belirlenmiştir.

> **Önemli kapsam uyarısı.** Bu skor, yalnızca **köşe-temelli (kenar artefaktı) kısayolunu** ölçer. §9'da gösterileceği üzere, omurga/mediasten gibi **merkez** bölgesine yerleşen bir kısayol bu metrikçe "düşük risk" olarak görünür — bu da neden köşe-skoru ile anatomik LFR'nin birlikte raporlanması gerektiğini açıklar.

```python
def compute_shortcut_score_v4(attn_map_14, attn_map_224=None, border_ratio=0.08):
    """
    v4 Kisayol Skoru — iki katmanlı analiz:

    Katman 1 — Patch-level (14×14):
      Doğrudan ViT patch token'larını karşılaştır.
      Köşe 4 patch (2×2) vs geri kalan 192 patch.
      → border_ratio sorunu yok, tam patch granülaritesi.

    Katman 2 — Piksel-level (224×224, opsiyonel):
      border_ratio=0.08 → %29 sınır alanı (makul).
      sadece 224px harita için kullanılır.

    Skor yorumu (v4):
      < 1.3 → Kose/merkez dengeli  🟢 DUSUK
      1.3-2.0 → Kose biraz yuksek  🟡 ORTA
      > 2.0 → Kose belirgin yuksek 🔴 YUKSEK (kisayol)

    Neden v3'ten daha iyi:
      v3: border_ratio=0.15 → %50 alan → median hep yakın → skor yakın
      v4: 14 patch-level → açık fark, anlamlı oran
    """
    # ── Patch-level analiz (birincil) ─────────────────────────────
    H14, W14 = attn_map_14.shape  # (14, 14)
    corner_mask_14 = np.zeros((H14, W14), dtype=bool)
    k = 2  # her köşeden 2×2=4 patch, toplam 16
    corner_mask_14[:k, :k]   = True
    corner_mask_14[:k, -k:]  = True
    corner_mask_14[-k:, :k]  = True
    corner_mask_14[-k:, -k:] = True

    corner_vals = attn_map_14[corner_mask_14]
    center_vals = attn_map_14[~corner_mask_14]

    corner_mean = float(corner_vals.mean())
    center_mean = float(center_vals.mean())
    score_patch = corner_mean / (center_mean + 1e-6)

    # ── 224px piksel-level analiz (ikincil) ───────────────────────
    if attn_map_224 is not None:
        H, W = attn_map_224.shape
        b = int(H * border_ratio)          # 0.08 → ~18px, %29 alan
        bm = np.zeros((H, W), dtype=bool)
        bm[:b,:]=True; bm[-b:,:]=True
        bm[:,:b]=True; bm[:,-b:]=True
        b_mean = float(attn_map_224[bm].mean())
        c_mean = float(attn_map_224[~bm].mean())
        score_pixel = b_mean / (c_mean + 1e-6)
    else:
        score_pixel = score_patch

    # Birincil skor: patch-level
    score = score_patch

    # İstatistik testi patch-level üzerinden
    stat, p_val = sp_stats.mannwhitneyu(
        corner_vals, center_vals, alternative='greater'
    )

    # v4 eşiği: 2.0 (haritalar daha yayılık → düşük zemin → oran daha ayrışık)
    shortcut = (score > 2.0) and (p_val < 0.05)

    if   score > 2.0: risk = 'YUKSEK'
    elif score > 1.3: risk = 'ORTA'
    else            : risk = 'DUSUK'

    return {
        'score'       : score,
        'score_patch' : score_patch,
        'score_pixel' : score_pixel,
        'corner'      : corner_mean,
        'center'      : center_mean,
        'p_value'     : float(p_val),
        'shortcut'    : shortcut,
        'risk'        : risk,
    }

# Dogrulama
print("Skor v4 dogrulanıyor...")
# Test 1: Merkez odaklı harita → dusuk skor bekleniyor
patch_center = np.zeros((14,14), dtype=np.float32)
patch_center[5:9, 5:9] = 1.0  # merkez parlak
patch_center = (patch_center-patch_center.min())/(patch_center.max()-patch_center.min()+1e-8)
sc_cen = compute_shortcut_score_v4(patch_center)
print(f"  Merkez odakli: skor={sc_cen['score']:.3f} → {'DUSUK bekleniyor' if sc_cen['score']<1.3 else 'SORUN'}")

# Test 2: Kose odaklı harita → yüksek skor bekleniyor
patch_corner = np.zeros((14,14), dtype=np.float32)
patch_corner[:2,:2]=1.0; patch_corner[:2,-2:]=1.0
patch_corner[-2:,:2]=1.0; patch_corner[-2:,-2:]=1.0
patch_corner = (patch_corner-patch_corner.min())/(patch_corner.max()-patch_corner.min()+1e-8)
sc_cor = compute_shortcut_score_v4(patch_corner)
print(f"  Kose odakli : skor={sc_cor['score']:.3f} → {'YUKSEK bekleniyor' if sc_cor['score']>2.0 else 'SORUN'}")

print(f"  Kisayol esigi v4: >2.0 (v3: >1.3)")
print("compute_shortcut_score_v4 hazir")


def compute_shortcut_score_v5(attn_map_14):
    """
    v5 Kalibreli Kisayol Skoru.
    Ham rollout (discard yok) uzerinden alan-agirlikli oran.
    Uniform dikkat → skor=1.0. Esik=1.55 (%5 yanlis pozitif).
    """
    total = attn_map_14.sum() + 1e-10
    attn_norm = attn_map_14 / total
    cm = np.zeros(attn_map_14.shape, dtype=bool)
    cm[:2,:2]=True; cm[:2,-2:]=True; cm[-2:,:2]=True; cm[-2:,-2:]=True
    n_corner, n_center = cm.sum(), (~cm).sum()
    corner_mean = float(attn_norm[cm].sum()  / n_corner)
    center_mean = float(attn_norm[~cm].sum() / n_center)
    score = corner_mean / (center_mean + 1e-10)
    from scipy import stats as _sp
    stat, p_val = _sp.mannwhitneyu(
        attn_norm[cm].flatten(), attn_norm[~cm].flatten(), alternative='greater')
    shortcut = (score > 1.55) and (float(p_val) < 0.05)
    risk = 'YUKSEK' if score > 1.6 else ('ORTA' if score > 1.3 else 'DUSUK')
    return {'score': score, 'score_patch': score, 'score_pixel': score,
            'corner': corner_mean, 'center': center_mean,
            'p_value': float(p_val), 'shortcut': shortcut, 'risk': risk}

print("compute_shortcut_score_v5 hazir (kalibrasyon: esik=1.55)")
```

## 9. Anatomik Akciğer-Odak Doğrulaması (Lung-Focus Ratio) <a id='10'></a>

Köşe skoru "modelin köşelere bakmadığını" söyler; ancak "modelin **akciğere baktığını**" garanti etmez. Bu boşluğu kapatmak için, dikkatin anatomik akciğer alanıyla örtüşmesini ölçen, doğrudan bir metrik tanımlanır.

### 9.1. Lung-Focus Ratio (LFR)

$S(u) \in [0,1]$ normalize edilmiş saliency (Attention Rollout) değeri, $L(u) \in \{0,1\}$ akciğer maskesi olsun. Düşük-enerjili arka plan gürültüsünü dışlamak için bir **gürültü eşiği** $\tau_n$ ile yüksek-enerjili piksel kümesi $\Omega = \{\,u : S(u) > \tau_n\,\}$ tanımlanır ($\tau_n = 0.2$). LFR, yüksek-enerjili saliency'nin akciğer içinde kalan oranıdır:

$$\boxed{\;\mathrm{LFR} \;=\; \frac{\displaystyle\sum_{u \in \Omega} S(u)\,L(u)}{\displaystyle\sum_{u \in \Omega} S(u)}\;}, \qquad \mathrm{LFR} \in [0,1].$$

$\mathrm{LFR} = 1$ tam akciğer-odaklı, $\mathrm{LFR} = 0$ tamamen akciğer-dışı dikkati ifade eder.

### 9.2. "Adil" LFR ve Bağımsız Segmentasyon

İki metodolojik titizlik uygulanır:

- **Bağımsız segmenter.** Akciğer maskesi, eğitimde kullanılan `lungmask`'tan *farklı* bir model (`ianpan/chest-x-ray-basic` U-Net) ile çıkarılır; böylece döngüsel kanıt önlenir. Maske yalnızca sağ + sol akciğeri içerir.
- **Adil maske (görünür akciğer).** Maske orijinal (kırpılmamış) görüntüde çıkarılır, ardından modelin gerçekte gördüğü uzaya, **aynı kırpma zinciriyle** ($\text{Resize}(256) \rightarrow \text{CenterCrop} \rightarrow \text{Resize}(224)$) taşınır. LFR yalnızca bu "görünür akciğer" üzerinde ölçülür; böylece modelin hiç göremediği (kırpılmış apeksler/kostofrenik açılar) haksız biçimde cezalandırılmaz.

Analiz, TRAIN / VAL / TEST bölümleri için ayrı ayrı sınıf-bazlı toplam (aggregate) ısı haritaları ve LFR özet istatistikleri üretir.

### 9.3. Bulgu

Ampirik sonuç, çalışmanın **en kritik bulgusudur:** Tüm bölümler genelinde küresel adil LFR yalnızca **$0.159 \pm 0.125$** (medyan $\approx 0.143$) düzeyindedir. Yani model köşelere bakmasa da, dikkatinin yalnızca ~%16'sı anatomik akciğer alanına düşmektedir. Toplam ısı haritaları, dikkatin akciğer loblarında değil, **iki akciğerin ortasındaki omurga/mediasten ekseninde** yoğunlaştığını gösterir. Bu, kenar kısayolundan daha sinsi, **merkez yerleşimli yeni bir kısayolun** açık kanıtıdır (yorumu §10'da).

```python
from scipy import ndimage as ndi
import cv2
import torchvision.transforms.functional as TF
from torchvision.transforms import InterpolationMode

# =====================================================================
# BÖLÜM 10 — ORİJİNAL (KIRPMA ÖNCESİ) GÖRÜNTÜ ÜZERİNDE AKCİĞER SEGMENTASYONU
# =====================================================================
# Önceki sürümde akciğer maskesi 224'lük KIRPILMIŞ görüntüde çıkarılıyordu.
# Artık segmentasyon kırpma ÖNCESİ orijinal görüntüde yapılır; böylece
# kostofrenik açılar / apeksler gibi kırpılan akciğer kısımları da görünür.
#
# Akış:
#   1) Orijinali yola göre yükle (gerekirse kısa kenar 768'e küçült)
#   2) ianpan seg modeliyle TÜM orijinalde akciğer maskesi çıkar
#   3) Modelin gerçekten gördüğü bölge = Resize(256)->CenterCrop(180)->Resize(224)
#      Bu geometrinin TERSİ ile attention'ı orijinal koordinata geri taşı
#   4) ADİL LFR: maske aynı kırpma zincirinden geçirilip 224'e indirgenir;
#      LFR yalnızca modelin görebildiği (kırpım içi) akciğer üzerinde ölçülür.

_SHORT_MAX = 768  # orijinalin kısa kenarını bu değere indirir (hız + figür çözünürlüğü)

def load_original_for_focus(path, short_max=_SHORT_MAX):
    """Orijinal görüntüyü (kırpma öncesi) yükler; çok büyükse kısa kenarı küçültür."""
    pil = Image.open(path).convert('RGB')
    W0, H0 = pil.size
    short = min(W0, H0)
    if short > short_max:
        sc = short_max / short
        W0, H0 = int(round(W0 * sc)), int(round(H0 * sc))
        pil = pil.resize((W0, H0), Image.BILINEAR)
    rgb     = np.asarray(pil).astype(np.float32) / 255.0   # (H0,W0,3) [0,1]
    gray_u8 = np.asarray(pil.convert('L'))                 # (H0,W0) uint8
    return rgb, gray_u8, (H0, W0)

def extract_lung_mask_gray(img_gray_u8, out_h, out_w):
    """ianpan modeliyle gri-seviye bir görüntüde akciğer maskesi (Sağ=1, Sol=2)."""
    x = seg_model.preprocess(img_gray_u8)
    x = torch.from_numpy(x).unsqueeze(0).unsqueeze(0).float().to(device)
    with torch.inference_mode():
        out = seg_model(x)
    mask_logits = out["mask"]
    # NOT: ianpan.preprocess kareye yeniden boyutlandırır; logit'i (H0,W0)'a
    # geri interpole etmek bu dönüşümü tersine çevirir (en-boy oranı korunur).
    mask_logits = F.interpolate(mask_logits, size=(out_h, out_w),
                                mode="bilinear", align_corners=False)
    pred = mask_logits.argmax(dim=1)[0].cpu().numpy()
    m = np.zeros_like(pred, dtype=np.float32)
    m[(pred == 1) | (pred == 2)] = 1.0
    m = cv2.GaussianBlur(m, (7, 7), 0)
    m = (m > 0.5).astype(np.float32)
    return m

# --- Kırpma geometrisi: 224 attention -> orijinal koordinat -------------
def _crop_box_in_original(H0, W0, resize_short=256, crop=None):
    """Resize(short)->CenterCrop(crop) sonrası kırpım kutusunu ORİJİNAL koordinatta verir."""
    if crop is None:
        crop = CONFIG["center_crop_size"]            # 180
    s = resize_short / min(H0, W0)                   # Resize(256) ölçek katsayısı
    RH, RW = int(round(H0 * s)), int(round(W0 * s))
    top_r  = int(round((RH - crop) / 2.0))
    left_r = int(round((RW - crop) / 2.0))
    top0  = top_r / s
    left0 = left_r / s
    side0 = crop / s
    return left0, top0, side0, s

def attn_to_original(attn_224, H0, W0):
    """224'lük attention'ı orijinal görüntü koordinatlarına geri taşır.
    Döndürür: (attn_orig HxW, analyzed_mask HxW bool, box=(left,top,side_px))."""
    left0, top0, side0, s = _crop_box_in_original(H0, W0)
    side_px = max(1, int(round(side0)))
    a = cv2.resize(attn_224.astype(np.float32), (side_px, side_px),
                   interpolation=cv2.INTER_CUBIC)
    canvas   = np.zeros((H0, W0), dtype=np.float32)
    analyzed = np.zeros((H0, W0), dtype=bool)
    t0 = max(0, int(round(top0)));  l0 = max(0, int(round(left0)))
    t1 = min(H0, t0 + side_px);     l1 = min(W0, l0 + side_px)
    canvas[t0:t1, l0:l1]   = a[:t1 - t0, :l1 - l0]
    analyzed[t0:t1, l0:l1] = True
    return canvas, analyzed, (l0, t0, side_px)

def lung_mask_orig_to_224(mask_orig):
    """Orijinal akciğer maskesini, modelin gördüğü 224 uzayına AYNI kırpma
    zinciriyle (nearest) taşır → ADİL LFR için 'görünür akciğer' maskesi."""
    m = torch.from_numpy(mask_orig.astype(np.float32))[None, None]        # (1,1,H0,W0)
    m = TF.resize(m, [256], interpolation=InterpolationMode.NEAREST)      # kısa kenar 256
    m = TF.center_crop(m, [CONFIG["center_crop_size"]] * 2)               # 180x180
    m = TF.resize(m, [CONFIG["img_size"]] * 2,
                  interpolation=InterpolationMode.NEAREST)                # 224x224
    return (m[0, 0].numpy() > 0.5).astype(np.float32)

# --- ADİL LFR (sadece KIRPIM İÇİNDEKİ akciğer üzerinde) -----------------
def lung_focus_ratio_filtered(saliency_map, lung_mask, noise_threshold=0.2):
    """Yüksek-enerjili (gürültü eşiği üstü) piksellerde akciğer-odak oranı.
    Maske 'görünür akciğer' (224 fair) olduğundan sonuç adildir."""
    saliency = (saliency_map.detach().cpu().numpy() if torch.is_tensor(saliency_map)
                else np.asarray(saliency_map)).flatten()
    mask = (lung_mask.detach().cpu().numpy() if torch.is_tensor(lung_mask)
            else np.asarray(lung_mask)).flatten()
    he = saliency > noise_threshold
    if not np.any(he):
        return float('nan')
    s, mk = saliency[he], mask[he]
    tot = s.sum()
    if tot < 1e-8:
        return float('nan')
    return float((s * mk).sum() / tot)

# --- Önizleme: ORİJİNAL (kırpma öncesi) + akciğer maskesi ---------------
print("Akciger maskesi (ORIJINAL goruntu) ornek dogrulama...")
_prev_paths = {}
for cls in CLASS_NAMES:
    idx = CLASS_TO_IDX[cls]
    _prev_paths[cls] = [p for p, l in image_datasets['test'].samples if l == idx][:2]

fig, axes = plt.subplots(2, 4, figsize=(14, 7))
fig.suptitle('Akciger Maskesi — ORIJINAL (kirpma oncesi) goruntu | ianpan SegModel',
             fontsize=12, fontweight='bold')
for r, cls in enumerate(CLASS_NAMES):
    for c in range(2):
        rgb, gray_u8, (H0, W0) = load_original_for_focus(_prev_paths[cls][c])
        mask = extract_lung_mask_gray(gray_u8, H0, W0)
        l0, t0, side0, _ = _crop_box_in_original(H0, W0)   # modelin gordugu kutu

        axes[r, c*2].imshow(rgb)
        axes[r, c*2].add_patch(patches.Rectangle((l0, t0), side0, side0,
                              lw=1.6, edgecolor='yellow', facecolor='none', ls='--'))
        axes[r, c*2].set_title(f'{cls} #{c+1} (orijinal)', fontsize=9, color=CLASS_COLORS[cls])
        axes[r, c*2].axis('off')

        ov = rgb.copy()
        ov[..., 0] = np.clip(ov[..., 0] + mask * 0.4, 0, 1)
        axes[r, c*2+1].imshow(ov)
        axes[r, c*2+1].contour(mask, levels=[0.5], colors='lime', linewidths=1.2)
        axes[r, c*2+1].set_title(f'Akciger (alan={mask.mean():.0%})', fontsize=9)
        axes[r, c*2+1].axis('off')
plt.tight_layout()
plt.savefig('fig_09_lung_mask_preview.png', dpi=120, bbox_inches='tight')
plt.show()
print("Orijinal-goruntu akciger segmentasyonu hazir.")
```

```python
# =====================================================================
# Aggregate Attention Rollout + ADİL Lung-Focus Ratio
#   - Maske ORİJİNALDE çıkarılır, modelin gördüğü 224 uzayına taşınır
#   - LFR yalnızca KIRPIM İÇİNDEKİ (görünür) akciğer üzerinde (adil)
#   - Aggregate ısı haritaları 224 uzayında tutulur (farklı orijinal
#     boyutlar toplanamaz); ADİL maskenin ortalaması kontur olarak çizilir.
# =====================================================================
model.eval()

MAX_PER_SPLIT = 40
splits        = ['train', 'val', 'test']

agg = {sp: {cls: {'cam_sum' : np.zeros((224, 224), dtype=np.float64),
                  'mask_sum': np.zeros((224, 224), dtype=np.float64),
                  'n': 0, 'lfr': [],
                  'n_empty_mask': 0, 'n_vanished': 0}
            for cls in CLASS_NAMES} for sp in splits}

def _iter_split_samples(sp):
    """(path, label) listesi — hem ImageFolder hem BalancedSubset '.samples' taşır."""
    return list(image_datasets[sp].samples)

# Hızlı sağlık kontrolü (orijinal segmentasyon)
print("Saglik kontrolu (orijinal segmentasyon ilk ornekler)...")
for cls in CLASS_NAMES:
    p = next(p for p, l in image_datasets['test'].samples if l == CLASS_TO_IDX[cls])
    _, g, (H0, W0) = load_original_for_focus(p)
    m = extract_lung_mask_gray(g, H0, W0)
    print(f"   {cls:<10} orijinal={H0}x{W0}  akciger alan={m.mean():.1%}")

for sp in splits:
    samples = _iter_split_samples(sp)
    random.Random(SEED).shuffle(samples)
    cnt = 0
    print(f'\n[{sp.upper()}] aggregate Rollout + ADIL LFR (max {MAX_PER_SPLIT})...')
    for path, lbl in samples:
        if cnt >= MAX_PER_SPLIT:
            break
        cls = IDX_TO_CLASS[lbl]
        try:
            # 1) Orijinali yükle + orijinalde akciğer segmentasyonu
            rgb, gray_u8, (H0, W0) = load_original_for_focus(path)
            lung_orig = extract_lung_mask_gray(gray_u8, H0, W0)

            # 2) Modelin gördüğü 224 uzayına 'görünür akciğer' maskesi
            lung_224 = lung_mask_orig_to_224(lung_orig)
            if lung_224.sum() < 1.0:
                agg[sp][cls]['n_empty_mask'] += 1
                continue

            # 3) Attention (model girişi: 'raw' transform = crop, maskesiz)
            img_t = data_transforms['raw'](Image.open(path).convert('RGB'))
            attn_224, _ = rollout.compute(img_t, discard_ratio=0.0)
            saliency = attn_224.astype(np.float32)
            sm, sM = saliency.min(), saliency.max()
            if sM - sm > 1e-8:
                saliency = (saliency - sm) / (sM - sm)

            # 4) ADİL LFR (yalnızca görünür akciğer)
            lfr = lung_focus_ratio_filtered(saliency, lung_224, noise_threshold=0.2)
            if np.isnan(lfr):
                agg[sp][cls]['n_vanished'] += 1
                continue

            agg[sp][cls]['cam_sum']  += saliency
            agg[sp][cls]['mask_sum'] += lung_224
            agg[sp][cls]['n']        += 1
            agg[sp][cls]['lfr'].append(lfr)
            cnt += 1
        except Exception as e:
            print(f'   hata ({os.path.basename(path)}): {e}')
            continue

    for cls in CLASS_NAMES:
        d = agg[sp][cls]
        if d['n'] > 0:
            extra = []
            if d['n_empty_mask']: extra.append(f"empty-mask={d['n_empty_mask']}")
            if d['n_vanished']:   extra.append(f"vanish={d['n_vanished']}")
            tag = f"  ({', '.join(extra)})" if extra else ""
            print(f'   {cls:<10}: n={d["n"]:>3}  ortalama ADIL LFR = {np.mean(d["lfr"]):.3f}{tag}')
        else:
            print(f'   {cls:<10}: gecerli ornek YOK '
                  f'(empty-mask={d["n_empty_mask"]}, vanish={d["n_vanished"]})')

print('\nAggregate (adil) hesap tamamlandi.')
```

### 9.4. Orijinal Koordinatlarda Birleşik Doğrulama <a id='10b'></a>

Önceki analizler dikkati modelin gördüğü $224 \times 224$ uzayda inceler. Bu bölüm, sonucu **klinik yoruma uygun orijinal görüntü düzlemine** geri taşır. Akciğer sınırları kırpılmamış orijinal radyografta çizilir; modelin baktığı bölgeler ise kırpma zincirinin **geometrik tersi** uygulanarak orijinal koordinatlara yansıtılır. Sarı kesikli kutu modelin fiilen gördüğü alanı, kutu dışındaki soluklaştırılmış bölge ise modelin hiç görmediği (kırpılmış) anatomiyi gösterir.

Bu birleşik gösterim iki şeyi aynı anda doğrular: (i) anatomik segmentasyonun orijinal ölçekte hatasız çalıştığını ve (ii) modelin dikkatinin, görebildiği akciğer alanı içinde dahi ısrarla santral (omurga/mediasten) bölgeye kaydığını. Böylece düşük LFR'nin bir kırpma artefaktı değil, modelin **içsel karar stratejisi** olduğu görsel olarak kanıtlanır.

```python
# =====================================================================
# fig_12 — BIRLESIK: ORIJINAL (kirpma oncesi) goruntu uzerinde
#          akciger bolgeleri + modelin tahmin icin baktigi bolgeler.
#          (Attention, kirpma zincirinin tersi ile orijinal koordinata tasindi.)
#
#   Sutun 1 : Orijinal X-ray + sari kesikli kutu (modelin GERCEKTEN gordugu bolge)
#   Sutun 2 : Orijinal + akciger bolgeleri (ianpan ile ORIJINALDE isaretli)
#   Sutun 3 : BIRLESIK -> akciger konturu + model dikkati (kutu icinde);
#             kutu disi soluklastirildi (model orayi hic gormedi) = durust gosterim
# =====================================================================
N_PER_CLASS = 3

sel = {}
for cls in CLASS_NAMES:
    paths = [p for p, l in image_datasets['test'].samples if l == CLASS_TO_IDX[cls]]
    random.Random(SEED + 1).shuffle(paths)
    sel[cls] = paths[:N_PER_CLASS]

n_rows = len(CLASS_NAMES) * N_PER_CLASS
fig, axes = plt.subplots(n_rows, 3, figsize=(13, 4.3 * n_rows))
if n_rows == 1:
    axes = axes[None, :]
fig.suptitle('Akciger Odak Dogrulamasi — ORIJINAL Goruntu Uzerinde (kirpma oncesi)\n'
             'Akciger bolgeleri + modelin baktigi bolgeler birlesik',
             fontsize=13, fontweight='bold', y=1.0)

row = 0
for cls in CLASS_NAMES:
    for path in sel[cls]:
        rgb, gray_u8, (H0, W0) = load_original_for_focus(path)
        lung_orig = extract_lung_mask_gray(gray_u8, H0, W0)

        # Attention (224) -> orijinal koordinat
        img_t = data_transforms['raw'](Image.open(path).convert('RGB'))
        attn_224, _ = rollout.compute(img_t, discard_ratio=0.0)
        a = attn_224.astype(np.float32)
        if a.max() - a.min() > 1e-8:
            a = (a - a.min()) / (a.max() - a.min())
        attn_orig, analyzed, (bl, bt, bs) = attn_to_original(a, H0, W0)

        # ADIL LFR (kirpim ici gorunur akciger)
        lung_224 = lung_mask_orig_to_224(lung_orig)
        lfr = lung_focus_ratio_filtered(a, lung_224, noise_threshold=0.2)

        # --- (1) Orijinal + kirpim kutusu ---
        ax = axes[row, 0]
        ax.imshow(rgb)
        ax.add_patch(patches.Rectangle((bl, bt), bs, bs, lw=1.8,
                     edgecolor='yellow', facecolor='none', ls='--'))
        ax.set_title(f'{cls} — Orijinal (kirpma oncesi)\nsari kutu = modelin gordugu',
                     fontsize=9, color=CLASS_COLORS[cls], fontweight='bold')
        ax.axis('off')

        # --- (2) Orijinal + akciger bolgeleri ---
        ax = axes[row, 1]
        ov = rgb.copy()
        ov[..., 1] = np.clip(ov[..., 1] + lung_orig * 0.35, 0, 1)   # yesil ton
        ax.imshow(ov)
        ax.contour(lung_orig, levels=[0.5], colors='lime', linewidths=1.4)
        ax.set_title('Akciger bolgeleri (orijinalde isaretli)', fontsize=9)
        ax.axis('off')

        # --- (3) Birlesik: akciger + model dikkati + dis soluk ---
        ax = axes[row, 2]
        base = rgb.copy()
        veil = np.where(analyzed[..., None], 1.0, 0.42)   # analiz disini karart
        base = (base * veil).clip(0, 1)
        ax.imshow(base)
        heat = np.ma.masked_where(~analyzed, attn_orig)   # isi haritasi yalnizca kutu icinde
        ax.imshow(heat, cmap=MEDICAL_CMAP, alpha=0.55, vmin=0, vmax=1)
        ax.contour(lung_orig, levels=[0.5], colors='lime', linewidths=1.4)
        ax.add_patch(patches.Rectangle((bl, bt), bs, bs, lw=1.6,
                     edgecolor='yellow', facecolor='none', ls='--'))
        lfr_txt = f'{lfr:.2f}' if not np.isnan(lfr) else 'NaN'
        bg = '#e8ffe8' if (not np.isnan(lfr) and lfr >= 0.65) else \
             ('#fff7e0' if (not np.isnan(lfr) and lfr >= 0.5) else '#ffe8e8')
        ax.set_title(f'Birlesik | ADIL LFR={lfr_txt}', fontsize=9, fontweight='bold',
                     bbox=dict(boxstyle='round,pad=0.25', facecolor=bg, edgecolor='#888'))
        ax.axis('off')
        row += 1

plt.tight_layout()
plt.savefig('fig_12_lung_focus_on_original.png', dpi=130, bbox_inches='tight')
plt.show()
print('fig_12 — orijinal koordinatli birlesik akciger-odak gorseli hazir.')
```

### 10.2. Temel Bulgu: Omurga/Mediasten Kısayolu

Test başarımı yüksektir (F1 = 0.9724, Doğruluk = %97.2) ve köşe kısayolu büyük ölçüde elenmiştir. Buna rağmen küresel adil $\mathrm{LFR} = 0.159 \pm 0.125$ değeri, modelin dikkatinin akciğer parankiminde değil, **iki akciğer arasındaki santral eksende (omurga + mediasten)** toplandığını gösterir. Bu örüntünün üç olası bilimsel nedeni belirlenmiştir:

1. **AP/PA Postür Yanlılığı.** Sağlıklı bireyler genellikle ayakta (PA) çekilirken, ağır pnömoni hastaları yatağa bağımlı (AP, taşınabilir cihaz) çekilir. Yatarak çekim geometrisinde kalp daha geniş, omurga daha belirgindir. Model, patolojiye bakmak yerine omurga yapısından hastanın **"yatan hasta (bedridden)"** olup olmadığını çıkarıp buradan haksız bir teşhis kısayolu üretiyor olabilir.
2. **Tıbbi Artefaktlar.** Göğüs ortasında ve omurga hizasında yer alan EKG kabloları, solunum cihazı hortumları gibi nesnelerin, model tarafından dolaylı bir "hastalık" göstergesi olarak ezberlendiği değerlendirilmektedir.
3. **Yetersiz Düzenlileştirme.** Anatomik ceza $\lambda = 5.0$ altında $\sim 3 \times 10^{-4}$ düzeyine inmektedir. Bu, modelin cezayı sayısal olarak küçültürken kararını hâlâ akciğer-dışı sinyale dayandırabildiğini, yani mevcut $\lambda$'nın bu güçlü kısayolu caydırmaya yetmediğini gösterir.

### 10.3. Çalışmanın Katkısı

Bu çalışmanın asıl bilimsel değeri, bir "%97 doğruluk" sonucu değil, **performans metriklerinin ardına gizlenmiş ciddi bir klinik yanlılığı niceliksel olarak gün yüzüne çıkaran bütünleşik bir denetim çerçevesidir.** Geliştirilen U-Net segmentasyonu, Attention Rollout ve adil LFR boru hattı, derin öğrenme modellerinin tıbbi teşhiste nasıl sinsi ve katmanlı kısayollar ürettiğini gösteren, yeniden kullanılabilir bir araç haline gelmiştir.

### 10.4. Gelecek Çalışmalar

- **$\lambda$ taraması ve programlanması:** $\lambda \in \{10, 20, 40\}$ ile artan agresiflik; F1–LFR ödünleşiminin Pareto cephesinin çıkarılması.
- **Postür dengeleme:** PA/AP meta-verisiyle katmanlı örnekleme veya domain-adversarial eğitim ile postür sinyalinin nedensel olarak dışlanması.
- **Pozitif yönlendirme:** Akciğer-dışını cezalandırmanın yanı sıra, radyolog-onaylı patoloji bölgelerine *pozitif* dikkat ödülü veren denetimli saliency hizalaması.
- **Çoklu-merkez doğrulama:** Modelin dağıtım kayması altındaki çöküşünü ölçmek için harici (dış-dağıtım) test kümeleri.

> **Klinik Not.** LFR klinik kabul eşiğinin (~0.65) belirgin altında olduğu ve santral kısayol giderilmediği sürece **klinik dağıtım önerilmez.** FDA/CE şeffaflık gereklilikleri uyarınca, dikkat haritalarının uzman radyolog onayından geçmesi zorunludur.
